```python
# ======================================================
# Notebook: 8D Hyperparameter Optimisation (Logistic Regression surrogate)
# Inputs: (40,8) | Output: (40,)
# Goal: maximise validation score
# ======================================================

import numpy as np
from sklearn.linear_model import LogisticRegression

# Load data
X = np.load("/mnt/data/initial_inputs.npy")      # (40,8)
y = np.load("/mnt/data/initial_outputs.npy")     # (40,)

# Convert to binary (top-performing region)
threshold = np.percentile(y, 75)
y_bin = (y >= threshold).astype(int)

# Train surrogate
model = LogisticRegression(max_iter=1000)
model.fit(X, y_bin)

# Candidate sampling
bounds = [(X[:,i].min(), X[:,i].max()) for i in range(8)]
n_candidates = 10000
X_grid = np.column_stack([
    np.random.uniform(b[0], b[1], n_candidates) for b in bounds
])

# Predict probability of high performance
probs = model.predict_proba(X_grid)[:,1]

# Select next (10,8)
top_idx = np.argsort(probs)[-10:]
next_points = X_grid[top_idx]

print("Next (10,8) hyperparameter candidates:")
print(next_points)
```
