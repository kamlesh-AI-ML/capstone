```python
# ======================================================
# Notebook: 6D Hyperparameter Optimisation (Logistic Regression surrogate)
# Inputs: (30,6) | Output: (30,)
# Goal: maximise performance
# ======================================================

import numpy as np
from sklearn.linear_model import LogisticRegression

# Load data
X = np.load("/mnt/data/initial_inputs.npy")      # (30,6)
y = np.load("/mnt/data/initial_outputs.npy")     # (30,)

# Binary target (top-performing region)
threshold = np.percentile(y, 75)
y_bin = (y >= threshold).astype(int)

# Train surrogate
model = LogisticRegression(max_iter=1000)
model.fit(X, y_bin)

# Candidate sampling
bounds = [(X[:,i].min(), X[:,i].max()) for i in range(6)]
n_candidates = 5000
X_grid = np.column_stack([
    np.random.uniform(b[0], b[1], n_candidates) for b in bounds
])

# Predict probability of best region
probs = model.predict_proba(X_grid)[:,1]

# Select next (10,6)
top_idx = np.argsort(probs)[-10:]
next_points = X_grid[top_idx]

print("Next (10,6) hyperparameter candidates:")
print(next_points)
```
