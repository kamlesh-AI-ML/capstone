```python
# ======================================================
# Notebook: 5D Recipe Optimisation (Logistic Regression surrogate)
# Inputs: (20,5) | Output: (20,)
# Goal: maximise (- total negative score)
# ======================================================

import numpy as np
from sklearn.linear_model import LogisticRegression

# Load data
X = np.load("/mnt/data/initial_inputs.npy")      # (20,5)
y_raw = np.load("/mnt/data/initial_outputs.npy") # (20,)

# Transform objective
y = -y_raw

# Binary target (top-performing recipes)
threshold = np.percentile(y, 75)
y_bin = (y >= threshold).astype(int)

# Train model
model = LogisticRegression(max_iter=1000)
model.fit(X, y_bin)

# Candidate sampling
bounds = [(X[:,i].min(), X[:,i].max()) for i in range(5)]
n_candidates = 5000
X_grid = np.column_stack([
    np.random.uniform(b[0], b[1], n_candidates) for b in bounds
])

# Predict probability of best region
probs = model.predict_proba(X_grid)[:,1]

# Select next (10,5)
top_idx = np.argsort(probs)[-10:]
next_points = X_grid[top_idx]

print("Next (10,5) candidate recipes:")
print(next_points)
```
