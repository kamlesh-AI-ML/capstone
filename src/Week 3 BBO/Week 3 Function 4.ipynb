```python
# ======================================================
# Notebook: Hyperparameter optimisation (SVM surrogate)
# Inputs: (30,4) | Output: (30,)
# Goal: minimise difference from baseline (maximize negative)
# ======================================================

import numpy as np
from sklearn.svm import SVR

# Load data
X = np.load("/mnt/data/initial_inputs.npy")      # (30,4)
y_raw = np.load("/mnt/data/initial_outputs.npy") # (30,)

# Transform objective (minimise -> maximise)
y = -y_raw

# Train SVM regression surrogate
model = SVR(kernel="rbf", C=10.0, gamma="scale")
model.fit(X, y)

# Generate candidate samples within bounds
bounds = [(X[:,i].min(), X[:,i].max()) for i in range(4)]
num_candidates = 5000
X_grid = np.column_stack([
    np.random.uniform(b[0], b[1], num_candidates) for b in bounds
])

# Predict surrogate score
preds = model.predict(X_grid)

# Exploration via distance from existing samples
dist = np.min(np.linalg.norm(X_grid[:,None,:] - X[None,:,:], axis=2), axis=1)

# Acquisition (exploitation + exploration)
acquisition = preds + 0.1 * dist

# Select next (10,4) candidates
top_idx = np.argsort(acquisition)[-10:]
next_points = X_grid[top_idx]

print("Next (10,4) hyperparameter candidates:")
print(next_points)
```
